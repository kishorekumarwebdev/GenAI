```python
# Step 1: Import Libraries

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input


# Step 2: Training Text (More Data)

text = """
Generative artificial intelligence is a branch of artificial intelligence.
It can create text, images, audio, videos, and other types of content.
Machine learning allows computers to learn patterns from data.
Deep learning uses neural networks to solve complex problems.
Neural networks are inspired by the human brain.
Large language models can understand and generate human language.
Artificial intelligence is transforming many industries.
Generative AI models are used in healthcare, education, business,
research, and software development.
Artificial intelligence learns from large amounts of data.
Machine learning enables intelligent systems.
Deep learning is a subset of machine learning.
Generative AI creates new content.
"""

text = text.lower()


# Step 3: Create Character Dictionary

chars = sorted(list(set(text)))

char_to_index = {c:i for i, c in enumerate(chars)}
index_to_char = {i:c for i, c in enumerate(chars)}

vocab_size = len(chars)

print("Number of characters:", vocab_size)


# Step 4: Prepare Training Data

sequence_length = 40

X = []
y = []

for i in range(len(text) - sequence_length):

    seq = text[i:i+sequence_length]
    target = text[i+sequence_length]

    X.append([char_to_index[c] for c in seq])
    y.append(char_to_index[target])


X = np.array(X)
y = np.array(y)


# Normalize Input

X = X / float(vocab_size)


# One Hot Encoding

y = tf.keras.utils.to_categorical(
    y,
    num_classes=vocab_size
)


# Reshape for LSTM

X = np.reshape(
    X,
    (X.shape[0], X.shape[1], 1)
)


# Step 5: Build LSTM Model

model = Sequential()

model.add(
    Input(shape=(sequence_length,1))
)

model.add(
    LSTM(128)
)

model.add(
    Dense(
        vocab_size,
        activation="softmax"
    )
)


model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)


model.summary()


# Step 6: Train Model

model.fit(
    X,
    y,
    epochs=200,
    batch_size=16,
    verbose=1
)


# Step 7: Text Generation Function

def generate_text(length=300, temperature=0.5):

    start = np.random.randint(
        0,
        len(X)-1
    )

    pattern = X[start]

    generated = ""

    for i in range(length):

        prediction = model.predict(
            pattern.reshape(1, sequence_length, 1),
            verbose=0
        )[0]


        # Temperature sampling
        prediction = np.log(prediction + 1e-8) / temperature
        prediction = np.exp(prediction) / np.sum(np.exp(prediction))


        index = np.random.choice(
            vocab_size,
            p=prediction
        )


        generated += index_to_char[index]


        pattern = np.append(
            pattern[1:],
            [[index / float(vocab_size)]],
            axis=0
        )


    return generated



# Step 8: Generate Output

print("\nGenerated Text:\n")

output = generate_text(
    length=300,
    temperature=0.4
)

print(output)
```
